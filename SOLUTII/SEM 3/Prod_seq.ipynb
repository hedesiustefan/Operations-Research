```python
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from Graphs import Node, Edge, FlowGraph
```

```python
time_m1 = [2, 3, 4, 2]      # minutes per unit on machine 1
time_m2 = [3, 2, 1, 2]      # minutes per unit on machine 2
capacity_m1, capacity_m2 = 500, 380
max_units = 100

source, m1, m2, sink = Node(name="s"), Node(name="M1"), Node(name="M2"), Node(name="t")
products = [Node(name=f"P{i + 1}") for i in range(4)]

edges = [Edge(origin=source, destination=m1, capacity=capacity_m1),
         Edge(origin=m2, destination=sink, capacity=capacity_m2)]
edges += [Edge(origin=m1, destination=p, capacity=max_units * time_m1[i]) for i, p in enumerate(products)]
edges += [Edge(origin=p, destination=m2, capacity=max_units * time_m2[i]) for i, p in enumerate(products)]

graph = FlowGraph([source, m1, m2, sink] + products, edges)
graph.max_flow("s", "t")
```

```python
pd.DataFrame([(u, v, d["capacity"], d["flow"]) for u, v, d in graph.edges(data=True)],
             columns=["from", "to", "capacity (min)", "flow (min)"])
```

```python
cut_value, (source_side, sink_side) = nx.minimum_cut(graph, "s", "t", capacity="capacity")
print("Minimum cut:", cut_value, "| sink side:", sink_side)
```
