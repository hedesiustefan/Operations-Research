## 3. Planning production in a plastic injection company

**Assumptions:** the machine uses one mould for one component per period, and delayed demand is carried over to the next period.

### a. Indexes and decision variables

**Indexes**
- $i$ = component type
- $m$ = mould
- $t$ = planning period ($t = 1, \dots, T$)

**Data**
- $h_i$ = storage cost per unit and period of component $i$
- $b_i$ = delay cost per unit and period of component $i$
- $I_{i0}$ = initial inventory of component $i$
- $I^{min}_i, I^{max}_i$ = minimum and maximum inventory of component $i$
- $C_{im}$ = units of component $i$ the machine can produce with mould $m$ in one period ($0$ if not possible)
- $d_{it}$ = demand of component $i$ in period $t$

**Decision variables**
- $x_{imt} \ge 0$ = units of component $i$ produced with mould $m$ in period $t$
- $I_{it} \ge 0$ = inventory of component $i$ at the end of period $t$
- $B_{it} \ge 0$ = units of component $i$ delayed at the end of period $t$
- $y_{imt} \in \{0, 1\}$ = 1 if the machine produces component $i$ with mould $m$ in period $t$

### b. Objective function

Minimise storage and delay costs:

$$\min \sum_{i} \sum_{t} \left( h_i I_{it} + b_i B_{it} \right)$$

### c. Constraints

**Inventory balance:** inventory and delays carried over, plus production, minus demand:
$$I_{i,t-1} - B_{i,t-1} + \sum_{m} x_{imt} - d_{it} = I_{it} - B_{it} \qquad \forall i, t \quad (B_{i0} = 0)$$

**Inventory limits:**
$$I^{min}_i \le I_{it} \le I^{max}_i \qquad \forall i, t$$

**Machine capacity:** production is only possible with a mould that is set up:
$$x_{imt} \le C_{im} \, y_{imt} \qquad \forall i, m, t$$

**One mould and component per period:**
$$\sum_{i} \sum_{m} y_{imt} \le 1 \qquad \forall t$$

### d. Set-up costs

New data: $S_m$ = set-up cost of mould $m$.

New variable: $s_{imt} \in \{0, 1\}$ = 1 if the machine is set up for component $i$ with mould $m$ in period $t$.

**Set-up detection:** a set-up happens when the machine makes something it did not make in the previous period:
$$s_{imt} \ge y_{imt} - y_{im,t-1} \qquad \forall i, m, t$$

**New objective function:**
$$\min \sum_{i} \sum_{t} \left( h_i I_{it} + b_i B_{it} \right) + \sum_{i} \sum_{m} \sum_{t} S_m \, s_{imt}$$
